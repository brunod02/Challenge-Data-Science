### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tiendas2 = pd.read_csv(url2)
tiendas3 = pd.read_csv(url3)
tiendas4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



### En este primer análisis, debes calcular el ingreso total de cada tienda. Sumando los valores de la columna Precio de cada conjunto de datos de la tienda para estimar los ingresos.

```python
tienda['Precio'].sum()
facturacion = tienda['Precio'].sum()
print(f'La facturación de la tienda 1 es de {facturacion}')
```

```python
tiendas2['Precio'].sum()
facturacion2 = tiendas2['Precio'].sum()
print(f'La facturación de la tienda 2 es de {facturacion2}')
```

```python
tiendas3['Precio'].sum()
facturacion3 = tiendas3['Precio'].sum()
print(f'La facturación de la tienda 3 es de {facturacion3}')
```

```python
tiendas4['Precio'].sum()
facturacion4 = tiendas4['Precio'].sum()
print(f'La facturación de la tienda 4 es de {facturacion4}')
```

# 2. Ventas por categoría

## En este debes calcular la cantidad de productos vendidos por categoría en cada tienda. La idea es agrupar los datos por categoría y contar el número de ventas de cada tipo, mostrando las categorías más populares de cada tienda.

```python
tienda.head()
```

```python
df_ventas_categorias1 = tienda['Categoría del Producto'].value_counts()
df_ventas_categorias1
```

```python
df_ventas_categorias1 = df_ventas_categorias1.reset_index()
df_ventas_categorias1 = df_ventas_categorias1.rename(columns={'count': 'Cantidad de Ventas'})
df_ventas_categorias = df_ventas_categorias1[df_ventas_categorias1['Cantidad de Ventas'] > 250]
df_ventas_categorias.plot(kind='line', x='Categoría del Producto', y='Cantidad de Ventas', color='green', figsize=(10,6))
```

```python
tiendas2.head()
```

```python
tienda2_ventas = tiendas2['Categoría del Producto'].value_counts()
tienda2 = tienda2_ventas.reset_index()
tienda2 = tienda2.rename(columns={'count': 'Cantidad de Ventas'})
tienda2 = tienda2[tienda2['Cantidad de Ventas'] > 250]
display(tienda2)
```

```python
df_ventas_categorias2 = tienda2
```

```python
df_ventas_categorias2.plot(kind='line', x='Categoría del Producto', y='Cantidad de Ventas', color='red', figsize=(10,6))
```
