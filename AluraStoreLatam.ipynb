# 4. Productos más y menos vendidos

```python
productos_mas1 = tienda['Producto'].value_counts().head(5)
productos_mas1
```

```python
productos_mas1 = tiendas2['Producto'].value_counts().head(5)
productos_mas1_df = productos_mas1.reset_index()
productos_mas1_df = productos_mas1_df.rename(columns={'count': 'Cantidad de Ventas'})
mas_vendido1 = productos_mas1_df
mas_vendido1
```

```python
productos_menos1 = tienda['Producto'].value_counts().tail(5)
productos_menos1_df = productos_menos1.reset_index()
productos_menos1_df = productos_menos1_df.rename(columns={'count': 'Cantidad de Ventas'})
menos_vendido1 = productos_menos1_df
menos_vendido1
```

```python
productos_totales = pd.concat([mas_vendido1,menos_vendido1])
```

```python
productos_totales.plot(kind='bar', x='Producto', y='Cantidad de Ventas', color='green')
```
