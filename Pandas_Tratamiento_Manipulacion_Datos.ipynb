# 1. Conociendo el problema

```python
import pandas as pd
```

```python
datos = pd.read_json('datos_hosting.json')
datos.head()
```

```python
datos = pd.json_normalize(datos['info_inmuebles'])
datos.head()
```

# 2. Datos numéricos

## Utilizando el metodo 'explode' para eliminar las listas dentro de las columnas

```python
columnas = list(datos.columns)
columnas
```

```python
datos = datos.explode(columnas[3:], ignore_index=True)
datos.head()
```
